```python
import os
import torch
import soundfile as sf
import numpy as np
from transformers import AutoModelForCausalLM
from IPython.display import Audio

model_id = "darkps/ice-012-audio"
text_prompt = "Hello! This voice agent is running completely free on a Kaggle cloud notebook."

print("Forcing model exclusively onto GPU 0...")
model = AutoModelForCausalLM.from_pretrained(
    model_id, 
    trust_remote_code=True, 
    dtype=torch.float16, 
    device_map={'': 'cuda:0'} 
)

print("Synthesizing audio...")
with torch.no_grad():
    audio_output = model.generate(
        text=text_prompt,
        language="English",
        gender="female"
    )

print("Processing and cleaning output array structure...")

# 1. Handle cases where the output is wrapped in a dictionary or has custom attributes
if isinstance(audio_output, dict):
    audio_data = audio_output.get("audio", audio_output.get("wav", audio_output))
else:
    audio_data = audio_output

# 2. Convert PyTorch tensors cleanly to CPU memory
if hasattr(audio_data, "cpu"):
    audio_data = audio_data.cpu().float().numpy()

# 3. Force the audio array to be a flat 1D sequence (Squeeze out batch or channel dimensions)
audio_data = np.squeeze(np.array(audio_data))

# 4. Fallback safeguard: If it is still multidimensional, extract the first audio channel stream
if audio_data.ndim > 1:
    audio_data = audio_data[0]

print(f"Final safe data shape for soundfile: {audio_data.shape}")
sampling_rate = 24000 

print("Saving playback file assets...")
sf.write("kaggle_output.wav", audio_data, sampling_rate)

print("🎉 Audio generation complete! Click play below:")
Audio("kaggle_output.wav")
```

```python
# 1. Install required packages
!pip install --quiet fastapi uvicorn pyngrok nest_asyncio transformers accelerate soundfile numpy pydub soxr
!pip install --quiet pyngrok nest_asyncio soxr

import os
import io
import torch
import soundfile as sf
import numpy as np
import nest_asyncio
from pyngrok import ngrok
from fastapi import FastAPI, HTTPException
from fastapi.middleware.cors import CORSMiddleware
from fastapi.responses import StreamingResponse
from pydantic import BaseModel
from transformers import AutoModelForCausalLM

from kaggle_secrets import UserSecretsClient
user_secrets = UserSecretsClient()
secret_value_0 = user_secrets.get_secret("NGROK_T")


# --- PASTE YOUR NGROK AUTH TOKEN HERE ---
NGROK_TOKEN = secret_value_0
ngrok.set_auth_token(NGROK_TOKEN)

app = FastAPI()

# Enable CORS so your local front-end website can query this Kaggle endpoint
app.add_middleware(
    CORSMiddleware,
    allow_origins=["*"],
    allow_credentials=True,
    allow_methods=["*"],
    allow_headers=["*"],
)

# Load the audio model directly onto the Kaggle cloud GPU memory
print("Loading model onto Kaggle GPU...")
model_id = "darkps/ice-012-audio"
model = AutoModelForCausalLM.from_pretrained(
    model_id, 
    trust_remote_code=True, 
    dtype=torch.float16, 
    device_map={'': 'cuda:0'}
)
print("🎉 Cloud Model Loaded Successfully!")

class TTSRequest(BaseModel):
    text: str
    language: str = "en"
    gender: str = "female"

@app.post("/v1/synthesize")
async def synthesize_speech(request: TTSRequest):
    try:
        with torch.no_grad():
            audio_output = model.generate(
                text=request.text,
                language=request.language,
                gender=request.gender
            )
        
        # Format normalization layers
        if isinstance(audio_output, dict):
            audio_data = audio_output.get("audio", audio_output.get("wav", audio_output))
        else:
            audio_data = audio_output

        if hasattr(audio_data, "cpu"):
            audio_data = audio_data.cpu().float().numpy()
            
        audio_data = np.squeeze(np.array(audio_data))
        
        buffer = io.BytesIO()
        sf.write(buffer, audio_data, 24000, format="WAV", subtype="PCM_16")
        buffer.seek(0)
        
        return StreamingResponse(buffer, media_type="audio/wav")
    except Exception as e:
        raise HTTPException(status_code=500, detail=str(e))

# 2. Open the public Internet Tunnel tunnel and fire up the Uvicorn engine
import uvicorn
port = 8000
public_url = ngrok.connect(port)
print(f"\n🚀 YOUR PUBLIC API ENDPOINT URL IS:\n{public_url.public_url}\n")

# nest_asyncio allows uvicorn to run seamlessly within a Jupyter Notebook kernel
nest_asyncio.apply()
import threading
import uvicorn
import nest_asyncio

# 1. Apply the nested async patch to allow multiple event loop references
nest_asyncio.apply()

# 2. Expose the server configuration layout
config = uvicorn.Config(
    app=app, 
    host="127.0.0.1", 
    port=8000, 
    loop="asyncio"
)
server = uvicorn.Server(config)

# 3. Define a worker thread to handle the blocking server logic in the background
def run_server():
    server.run()

# 4. Fire up the background thread
server_thread = threading.Thread(target=run_server, daemon=True)
server_thread.start()

print("\n🔥 BACKGROUND ENDPOINT IS NOW RUNNING!")
print(f"👉 Your web application can now hit: https://ngrok-free.dev\n")
print("⚠️ DO NOT stop this Kaggle cell, or the API server thread will terminate.")

```
